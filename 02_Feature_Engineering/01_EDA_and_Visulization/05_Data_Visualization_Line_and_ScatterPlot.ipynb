```python
# from a pandas dataframe
batsman = pd.read_csv('sharma-kohli.csv')
batsman
```

```python
# from a pandas dataframe

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='solid',linewidth=2,marker='+')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='solid',linewidth=3,marker='D',markersize=10,label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='solid',linewidth=2,marker='+',label='Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparision')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc='upper right')
plt.grid()
plt.show()
```

# Bar chart

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='skyblue')
```

```python
# horizontal bar chart
plt.barh(colors,children,color='skyblue')
```

```python
# color and lebel
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
plt.bar(df['batsman'],df['2017'])
```

```python
plt.bar(df['batsman'],df['2015'],width=.5)
```

```python
plt.bar(np.arange(df.shape[0]),df['2015'],width=0.5)
```

```python

plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='pink')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='skyblue')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='gray')

plt.xticks(np.arange(df.shape[0]),df['batsman'])

plt.show()
```

```python
# stacked bar chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=df['2016'] + df['2017'],label='2015')

plt.legend()
plt.show()
```
