# Web Scraping using BeautifulSoup
### Project: Extracting Top companies Data from AmbitionBox

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
requests.get('https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav').text
```

```python
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/107.0.0.0 Safari/537.36'}
webpage=requests.get('https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav', headers=headers).text
```

```python
soup=BeautifulSoup(webpage, 'html.parser')
```

## Target Identification & Loop Logic

```python
companies = soup.find_all('h2')

# 2. Pehli 5 companies ke naam print karne ke liye loop
for i in companies[:5]:
    print(i.text.strip())
```

## Structured DataFrame & CSV Expert

```python
names_list = [i.text.strip() for i in soup.find_all('h2')]

# 2. Is list se clean data table (DataFrame) banana
df_final = pd.DataFrame({'Company Name': names_list})

# 3. Table ko display karne ke liye (Sabse zaroori)
df_final.head()
```

```python
df_final.to_csv('companies.csv',index=False)
```
