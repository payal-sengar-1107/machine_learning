```python
import json
import pandas as pd

# 1. real-world sample dataset hum direct memory mein rakh rahe hain
data_json = """
[
  {"id": 1, "cuisine": "greek", "ingredients": ["romaine lettuce", "black olives", "grape tomatoes"]},
  {"id": 2, "cuisine": "southern_us", "ingredients": ["plain flour", "ground pepper", "salt", "tomatoes"]},
  {"id": 3, "cuisine": "filipino", "ingredients": ["eggs", "pepper", "salt", "mayonnaise"]}
]
"""

# 2. JSON data ko Python formats mein convert karna
clean_data = json.loads(data_json)

# 3. Pandas DataFrame (Table) banana, 
df_movies = pd.DataFrame(clean_data)

# 4. Final Table ka preview dekhne ke liye
df_movies.head()
```
